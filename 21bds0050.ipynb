```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("Gunnels.csv")
df.head()
```

```python
# 1. Explore Dimensions & Summary
print("Dataset Dimensions:", df.shape)
print("\nDataset Info:")
df.info()
print("\nSummary Statistics:")
print(df.describe())
```

```python
# 2. Handle Missing Values
print("\nMissing Values:")
print(df.isnull().sum())
df.fillna(df.mean(), inplace=True)  # Impute missing values with mean
```

```python
# 3. Data Cleaning (Removing Duplicates, Checking Data Types)
df.drop_duplicates(inplace=True)
df.columns = df.columns.str.strip()  # Remove spaces in column names
```

```python
# 4. Univariate Analysis
for column in df.select_dtypes(include=['number']).columns:
    plt.figure(figsize=(12, 5))
    sns.histplot(df[column], kde=True)
    plt.title(f"21BDS0050 Univariate Analysis - Histogram of {column}")
    plt.show()

    plt.figure(figsize=(12, 5))
    sns.boxplot(x=df[column])
    plt.title(f"21BDS0050 Univariate Analysis - Boxplot of {column}")
    plt.show()
```

```python
# 5. Bivariate Analysis
numeric_cols = df.select_dtypes(include=['number']).columns
for i in range(len(numeric_cols)):
    for j in range(i+1, len(numeric_cols)):
        plt.figure(figsize=(8, 6))
        sns.scatterplot(x=df[numeric_cols[i]], y=df[numeric_cols[j]])
        plt.title(f"21BDS0050 Bivariate Analysis - Scatter Plot: {numeric_cols[i]} vs {numeric_cols[j]}")
        plt.show()

        plt.figure(figsize=(8, 6))
        sns.boxplot(x=df[numeric_cols[i]], y=df[numeric_cols[j]])
        plt.title(f"21BDS0050 Bivariate Analysis - Boxplot: {numeric_cols[i]} vs {numeric_cols[j]}")
        plt.show()

        plt.figure(figsize=(8, 6))
        sns.regplot(x=df[numeric_cols[i]], y=df[numeric_cols[j]])
        plt.title(f"21BDS0050 Bivariate Analysis - Regression Plot: {numeric_cols[i]} vs {numeric_cols[j]}")
        plt.show()
```

```python
# 6. Multivariate Analysis
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("21BDS0050 Multivariate Analysis - Correlation Heatmap")
plt.show()

sns.pairplot(df)
plt.title("21BDS0050 Multivariate Analysis - Pairplot")
plt.show()
```

#APPLYING TIME SERIES ANALYSIS IN MY DATASET


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA
from datetime import datetime

print(df.info())
print(df.head())

plt.figure(figsize=(12,5))
plt.plot(df, label="Time Series Data")
plt.xlabel("Date")
plt.ylabel("Value")
plt.title("21BDS0050 Time Series Visualization")
plt.legend()
plt.show()

print("Missing values:\n", df.isnull().sum())

df.fillna(method='ffill', inplace=True)

decomposition = seasonal_decompose(df[df.columns[0]], model='additive', period=12)
decomposition.plot()
plt.show()

p, d, q = 1, 1, 1
model = ARIMA(df[df.columns[0]], order=(p, d, q))
model_fit = model.fit()
print(model_fit.summary())


forecast_steps = 30
forecast = model_fit.forecast(steps=forecast_steps)
forecast_index = pd.date_range(start=df.index[-1], periods=forecast_steps+1, freq='D')[1:]

plt.figure(figsize=(12,5))
plt.plot(df[df.columns[0]], label="Actual Data")
plt.plot(forecast_index, forecast, label="Forecast", linestyle='dashed')
plt.xlabel("Date")
plt.ylabel("Value")
plt.title("21BDS0050 Time Series Forecasting with ARIMA")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
import warnings
warnings.filterwarnings("ignore")
```

```python
# ===================== [ STEP 2: DATA CLEANING ] =====================
df = df.drop(columns=["rownames"])  # Drop ID/index column if not useful
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nData Types:\n", df.dtypes)
```

```python
# ===================== [ STEP 3: DATA HANDLING ] =====================
X = df.drop(columns=["Gunnel"])
y = df["Gunnel"]

# Normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

```

```python
# ===================== [ STEP 4: 1D, 2D, N-D VISUALIZATION ] =====================
# 1D: Histogram & Boxplot
for col in X.columns:
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    sns.histplot(X_scaled_df[col], kde=True, color='skyblue')
    plt.title(f'Histogram of {col}')
    plt.subplot(1, 2, 2)
    sns.boxplot(x=X_scaled_df[col], color='lightgreen')
    plt.title(f'Boxplot of {col}')
    plt.tight_layout()
    plt.show()
```

```python
# 2D: Correlation Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(X_scaled_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# N-D: Pairplot (top 4 features + target)
temp = X_scaled_df.copy()
temp['Gunnel'] = y
sns.pairplot(temp[['Time', 'Fromlow', 'Slope', 'Rw', 'Gunnel']], hue='Gunnel', palette='husl')
plt.suptitle("Pairplot", y=1.02)
plt.show()
```

```python
# ===================== [ STEP 5: TIME SERIES ANALYSIS ] =====================
# Check time trend if 'Time' represents a timestamp
plt.figure(figsize=(12, 5))
sns.lineplot(x=df["Time"], y=df["Gunnel"])
plt.title("Time vs Gunnel Occurrence")
plt.xlabel("Time")
plt.ylabel("Gunnel (Presence=1, Absence=0)")
plt.grid()
plt.show()
```

```python
# ===================== [ STEP 6: ANALYSIS OF 1D, 2D, N-D DATA ] =====================
# Descriptive stats
print("\nDescriptive Stats:\n", df.describe())

# Grouped analysis
print("\nGrouped by Gunnel (mean values):\n", df.groupby('Gunnel').mean())
```

```python
# ===================== [ STEP 7: DIMENSIONALITY REDUCTION ] =====================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y, palette="Set1")
plt.title("PCA Projection (2D)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)
plt.show()
```

```python
# ===================== [ STEP 8: MODEL BUILDING & EVALUATION ] =====================
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Evaluation
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
